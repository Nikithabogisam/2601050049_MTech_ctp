```python
students = []

n = int(input("Enter number of students: "))

for i in range(n):
    print("\nEnter details of student", i + 1)

    name = input("Enter student name: ")
    total_classes = int(input("Enter total classes conducted: "))
    attended_classes = int(input("Enter classes attended: "))

    attendance = (attended_classes / total_classes) * 100

    student = {
        "name": name,
        "total": total_classes,
        "attended": attended_classes,
        "attendance": attendance
    }

    students.append(student)

print("\n--- STUDENT ATTENDANCE DETAILS ---")

for student in students:
    print("\nName:", student["name"])
    print("Total classes:", student["total"])
    print("Classes attended:", student["attended"])
    print("Attendance:", round(student["attendance"], 2), "%")

    if student["attendance"] < 75:
        print("Status: Below 75%")

print("\n--- STUDENTS BELOW 75% ---")

found = False

for student in students:
    if student["attendance"] < 75:
        print(student["name"], "-", round(student["attendance"], 2), "%")
        found = True

if found == False:
    print("No student is below 75%.")

highest = students[0]

for student in students:
    if student["attendance"] > highest["attendance"]:
        highest = student

print("\n--- HIGHEST ATTENDANCE ---")
print("Student:", highest["name"])
print("Attendance:", round(highest["attendance"], 2), "%")

total_attendance = 0

for student in students:
    total_attendance = total_attendance + student["attendance"]

average = total_attendance / len(students)

print("\nClass average attendance:", round(average, 2), "%")
```
