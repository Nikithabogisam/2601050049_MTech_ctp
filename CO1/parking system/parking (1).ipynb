```python
total_slots = 100
parking = {}

while True:
    print("\n--- PARKING SYSTEM ---")
    print("1. Display available slots")
    print("2. Vehicle Entry")
    print("3. Vehicle Exit")
    print("4. Exit Program")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        available = total_slots - len(parking)
        print("Available parking slots:", available)

        if available == 0:
            print("Parking area is FULL")

    elif choice == 2:
        if len(parking) == total_slots:
            print("Parking area is FULL")
        else:
            vehicle_number = input("Enter vehicle number: ")

            if vehicle_number in parking:
                print("Vehicle is already parked.")
            else:
                for slot in range(1, total_slots + 1):
                    if slot not in parking.values():
                        break

                entry_time = int(input("Enter entry time (24 hour format): "))

                parking[vehicle_number] = slot, entry_time

                print("Vehicle parked successfully.")
                print("Your parking slot is:", slot)

    elif choice == 3:
        vehicle_number = input("Enter vehicle number: ")

        if vehicle_number in parking:
            slot, entry_time = parking[vehicle_number]

            exit_time = int(input("Enter exit time (24 hour format): "))

            duration = exit_time - entry_time

            if duration <= 0:
                duration = 1
                
            fee = duration * 20

            print("Parking slot:", slot)
            print("Parking duration:", duration, "hour(s)")
            print("Parking fee: Rs.", fee)

            del parking[vehicle_number]

            print("Vehicle exited successfully.")

        else:
            print("Vehicle not found.")

    elif choice == 4:
        print("Thank you!")
        break

    else:
        print("Invalid choice.")
```
